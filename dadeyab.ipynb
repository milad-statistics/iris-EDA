3. Handling Missing Values

Missing values were handled using group-wise median imputation.

Instead of calculating a single median for the entire dataset, the median of each numerical feature was calculated separately for each Iris species using groupby() and transform().

The missing values were then replaced with the corresponding species-specific median using fillna().

This approach preserves the differences between species and avoids using a global median that may not accurately represent each species' measurements.

```python
plt.figure(figsize=(15,10))
plt.subplot(2,2,1)
plt.title('Histogram of PetalLengthCm')
sns.histplot(df.PetalLengthCm,kde=True)
plt.axvline(df.PetalLengthCm.mean(),color='red',label='mean')
plt.legend()
plt.subplot(2,2,2)
plt.title('Histogram of PetalWidthCm')
sns.histplot(df.PetalWidthCm,kde=True)
plt.axvline(df.PetalWidthCm.mean(),color='red',label='mean')
plt.legend()
plt.subplot(2,2,3)
plt.title('Histogram of SepalLengthCm')
sns.histplot(df.SepalLengthCm,kde=True)
plt.axvline(df.SepalLengthCm.mean(),color='red',label='mean')
plt.legend()
plt.subplot(2,2,4)
plt.title('Histogram of SepalWidthCm')
sns.histplot(df.SepalWidthCm,kde=True)
plt.axvline(df.SepalWidthCm.mean(),color='red',label='mean')
plt.legend()
plt.suptitle('Histogram of all the variables pooled')
plt.tight_layout()
```

```python
# Q:which numerical features are most effective in distinguishing the three iris species?
F = ["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"]
for f in F:
    H, p = stats.kruskal(*[g[f].dropna() for _, g in df.groupby("Species")])
    print(f"{f}: H = {H:.1f}, p = {p:.1e}")
```

4. Histograms of Numerical Features

Four histograms were created to examine the distributions of the numerical features: SepalLengthCm, SepalWidthCm, PetalLengthCm, and PetalWidthCm.

The sns.histplot() function was used with kde=True to display the histograms along with their Kernel Density Estimation (KDE) curves. Additionally, a red vertical line was added to each histogram using plt.axvline() to indicate the mean of the corresponding feature. Legends were included to identify the mean lines.

These visualizations provide an overview of the distribution and central tendency of each numerical feature.

Comparing Numerical Features Across Species Using the Kruskal–Wallis H Test

The Kruskal–Wallis H test was performed separately on all four numerical features to examine differences among the three Iris species.

The test results showed that all four features had extremely small p-values (all below 0.05), indicating statistically significant differences in their distributions across the species.

Among the four features, PetalWidthCm produced the highest H-statistic, approximately 128.

Therefore, PetalWidthCm showed the strongest statistical evidence of differences among the three Iris species in this comparison.
